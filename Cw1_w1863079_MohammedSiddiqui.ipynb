## Data Mining in Car Insurance

In the auto insurance sector, data mining focusses on examining huge databases to find insightful information. A better grasp of consumer behaviour, risk levels, claims patterns, and fraudulent activity is made possible by these insights. Analysis of consumer behaviour enables insurers to provide individualised insurance, enhance client retention, and customise services. More precise premium calculations based on driving history, vehicle type, and accident probabilities are made possible by risk assessment models. Trends, including seasonal claim spikes or recurrent accident causes, can be found by looking for patterns in claims data.

One crucial use case is fraud detection, where data mining tools find questionable claim patterns like inflated damages or duplicate claims. Insurance companies can lower fraudulent payouts and safeguard their profits by utilising machine learning and anomaly detection technologies. All things considered, data mining improves customer service, operational effectiveness, and decision-making in the auto insurance industry by fostering more proactive and data-driven procedures.

### Key Areas in Car Insurance Data Mining

- **Fraud Detection**: Examining data for anomalies, such as inconsistent reported incidents, unusually frequent claims, inflated damage reports, duplicate claims across policies, and correlations between particular policyholders and high-risk behaviours, is necessary to identify patterns that point to fraudulent claims. Machine learning models, clustering algorithms, and anomaly detection techniques are then used to flag suspicious activity and assist proactive fraud prevention.

- **Customer Segmentation**: By using clustering algorithms and segmentation techniques to create distinct customer profiles, insurers can group customers based on factors like driving behaviour, claim history, vehicle type, and demographic information. This allows insurers to customise insurance products, optimise pricing strategies, and enhance customer engagement by providing coverage options that correspond with individual needs and risk levels.

- **Pricing Models**: Using data-driven algorithms to evaluate client risk profiles, past claim trends, and anticipated claim costs, insurers may establish premiums that reflect individual risk levels while maintaining competitiveness, profitability, and customer justice. They can create pricing strategies based on risk assessment and claim forecasts thanks to this procedure.

- - **Claim Prediction**: By using predictive modelling techniques, such as regression analysis and machine learning algorithms, to analyse past claims, driving records, vehicle details, and customer demographics, insurers can estimate the likelihood of future claims and forecast potential costs, supporting more accurate premium pricing and financial risk management. This process predicts the likelihood and cost of claims based on historical data.

- **Risk Assessment**: In order to evaluate client risk profiles, data mining techniques like regression analysis, decision trees, and machine learning models are used to classify clients into risk categories. This allows insurers to determine fair premiums, identify high-risk behaviours, and provide customised insurance plans that balance profitability and customer satisfaction. Other factors that are analysed include driving history, including past accidents and traffic violations; vehicle type; repair costs and safety ratings; and demographic information like age, location, and occupation.


### AI Techniques in Data Mining for Car Insurance

#### Linear Regression

- **Used for**: In order to estimate potential claim expenses and determine the overall risk level, machine learning models and statistical techniques are used to analyse factors like driving history, claim frequency, vehicle specifications, and demographic information. This allows insurers to make well-informed decisions regarding premium pricing and policy terms.
- **Limitation**: Complex or non-linear patterns in the data may not be adequately captured by linear regression, which makes the assumption that variables have a linear relationship.
- **Strength**: By fitting a straight line to estimate the relationship between input variables and the desired outcome, linear regression is a straightforward and interpretable model for linear relationships that makes it simple to comprehend how changes in each variable impact the anticipated outcome.

#### Random Forest

- **Ideal for**: By analysing customer data and identifying patterns that assign risk levels or flag suspicious activities for additional investigation, classification algorithms like decision trees, support vector machines, or neural networks are used to classify customers into various risk segments or detect fraudulent claims.
- **Limitation**: Because random forests integrate several decision trees, it is challenging to track the process by which individual predictions are formed, making them less interpretable.
- **Strength**: By merging several decision trees and calculating the contribution of each feature to the forecast accuracy of the model, random forests manage intricate, non-linear interactions and offer insights on feature importance.


#### Logistic Regression 

- **Ideal for**: Classification models such as logistic regression, decision trees, or neural networks are used to assess the probability of each occurrence and categorise the result based on a predetermined threshold in order to predict binary outcomes, such as claim approvals or denials.
- **Limitation**: The capacity of logistic regression to identify intricate feature interactions and non-linear patterns may be limited by the assumption of a linear relationship between features and the log-odds of the result.
- **Strength**: By calculating the likelihood of an outcome based on input features, logistic regression is a rapid, easy-to-understand model that may be used for binary classification.

## Technique 1: Linear Regression

A popular statistical method for calculating the projected claim amount based on variables including client demographics, vehicle specs, and past claim history is linear regression, which is essential for forecasting the costs of auto insurance claims. Important factors to take into account while using linear regression in this situation include expected results, setup and processing times, and data availability. A complete dataset with both numerical and categorical characteristics pertinent to claims, such as driver age, vehicle type, and prior claims history, is needed by the model. Its efficiency enables it to generate findings quickly, even with huge datasets, and its straightforward structure and low hyperparameter adjustment make it suited for real-time research. A linear regression model's output includes feature coefficients that show how each variable affects the prediction as well as a continuous number that represents the expected claim cost. The model may have drawbacks when handling complicated interactions or non-linear correlations in the data, despite its interpretability and simplicity.


#### Data Availability

To make precise forecasts, linear regression uses historical data. Insurance companies must have broad access to a variety of historical datasets that contain important information about clients and their interactions in order to process claims for auto insurance. Client demographics, including age and gender, car details, like make and model, driving histories, insurance claims, and any pertinent background data are usually included in these datasets. By examining these factors, the model finds trends and creates connections between the characteristics of the consumer and the car and the related claim expenses. According to Hastie, Tibshirani, and Friedman (2009), this procedure helps insurers to more accurately predict future claims and optimise their pricing strategies based on data-driven insights. A strong and trustworthy model must be constructed using thorough historical records because incomplete or skewed data can lower the model's predicted accuracy.

#### Mean Squared Error (MSE)

In regression analysis, the Mean Squared Error (MSE) is a commonly used metric to assess a prediction model's performance and accuracy. It calculates the mean squared discrepancy between the observed data and the model's matching predictions. Better accuracy is indicated by a lower MSE, which shows that the model's predictions are more in line with the actual values. MSE is useful for emphasising notable deviations because it penalises greater errors more severely by squaring the differences. Finding the best-fitting model for accurate predictions is made easier with the use of this metric, which offers a simple means of comparing model performance during training and validation.

#### Interpretation

A thorough indicator of how closely the model's predictions match the actual data is the MSE number. Stronger model performance and more prediction accuracy are demonstrated by a lower MSE, which shows that the projected values closely match the observed data. A greater MSE, on the other hand, indicates a wider discrepancy between the expected and actual values, suggesting subpar model performance and other problems like underfitting or inadequate feature selection. The MSE aids in evaluating model reliability and directs enhancements to model tuning and design by measuring prediction errors.

#### Time to Setup

The comparatively short setup time of linear regression is one of its main benefits. Depending on the dataset's complexity, creating a linear regression model to forecast auto insurance claims can frequently be finished in a few days. Essential steps in the setup process include feature engineering to choose and modify pertinent variables, data preparation to deal with missing values and outliers, and model selection and validation to guarantee dependable performance. Even though these procedures must be carefully followed, Linear Regression is known for its simple structure and low processing requirements, which enable rapid deployment. Because of its effectiveness, it is a sensible option for insurers who want to create predictive models without having to deal with long implementation periods (Hastie et al., 2009).

#### Time to Produce Results

In the ever-changing world of auto insurance claims, Linear Regression models are extremely important because, if put into practice, they can generate forecasts in real time. Insurers frequently encounter circumstances that call for prompt risk evaluations and premium changes. Because the model can provide quick cost estimates, insurers may quickly assess risk, determine suitable premiums, and make smart underwriting choices. The total effectiveness of claims management procedures is increased by this responsiveness and speed, which guarantees prompt actions that boost operational efficacy and customer service. One of the main reasons why linear regression is so often used to facilitate quick decisions in cutthroat insurance markets is its real-time prediction power (Paczolt, 2023).

#### Output

A numerical estimate of the anticipated claim cost is the main output of a linear regression model when it comes to auto insurance claims. A cost prediction that represents the expected financial obligation is produced by the model given a set of input features, including car specifications, customer demographics, and other pertinent information. These forecasts are used by insurers to inform underwriting choices and establish premium rates that correspond with each policyholder's evaluated risk profile. The numerical output helps insurers maintain competitive pricing strategies, manage risk more efficiently, and guarantee fairness in premium calculations by offering a clear and data-driven estimate. This predictive ability helps make data-driven decisions and makes insurance procedures more precise and open (Paczolt, 2023).

#### Weaknesses

- **Linearity Assumption:** Complex interactions or non-linear patterns in the data may not be adequately captured by linear regression, which presupposes a linear relationship between the input parameters and the target variable, such as claim cost. In situations when the interactions between variables are more complex or impacted by numerous interrelated factors, this constraint may result in poorer prediction accuracy.
- **Limited Complexity:** When variables affect the objective in more complicated ways or when patterns depart from a straight-line trend, linear regression may not be as useful because it may miss complex interactions or non-linear relationships in the data.
- **Data Requirements:** For insurers with limited data access or those entering new markets where adequate historical records may not be accessible, Linear Regression is less successful because it mainly depends on historical data to find trends and make forecasts. In these situations, the model's accuracy and dependability may be seriously jeopardised because there is not enough training data.

#### Strengths

- **Predictive Accuracy:** A helpful method for calculating the anticipated cost of insurance claims depending on a number of variables is linear regression. It works best when there are roughly linear correlations between claim costs and input data, such as customer demographics, vehicle characteristics, and claim history. The model helps insurers understand how various attributes contribute to claim expenses by fitting a linear equation to the data and producing predictions that are easy to interpret. It may perform poorly in situations when the underlying relationships in the data are complex or non-linear, though, as its accuracy is dependent on the assumption of linearity.
- **Quick Setup:** For insurers who require quick risk assessments, linear regression models are a great option because they are easy to deploy. Quick data preparation, feature selection, and model training are made possible by their simple structure and low processing demands. This quick setup makes it possible for insurers to quickly use predictive models, facilitating quicker decision-making in dynamic settings where quick risk assessments and premium changes are crucial.
- **Real-time Predictions:** Insurers can swiftly evaluate risk and modify premiums as necessary thanks to linear regression's ability to produce forecasts in real time. This responsiveness is essential in dynamic situations where prompt risk assessments and policy modifications can boost customer happiness, increase operational effectiveness, and facilitate well-informed underwriting choices. The model is a useful tool for sustaining competitive and flexible pricing strategies because of how quickly it can produce forecasts.

#### Advantages and Disadvantages

The main benefits of linear regression are its ease of use, interpretability, and speed of implementation, which make it simple to comprehend and apply. Its primary flaw, though, is the simplistic assumption that the connection between the target variable and the input features is linear. ## Improvements

**Addressing Class Imbalance**
- There is probably a large disparity between fraudulent and non-fraudulent claims in our dataset, with the former being much more common. Models may become biassed towards predicting the majority class as a result of this imbalance, which could lead to inadequate detection of false claims. We can use strategies like undersampling the majority class to balance the dataset, oversampling the minority class to increase the representation of fraudulent claims, or class-weight adjustments during model training or algorithms like SMOTE (Synthetic Minority Over-sampling Technique) that are made to deal with imbalanced data. These methods enhance the model's capacity to spot false assertions without sacrificing overall functionality.:

# TEST 2: Oversampling the Minority Class

```python
# Preparing the dataset: Encoding the target variable and one-hot encoding categorical variables
data['fraud_reported'] = data['fraud_reported'].apply(lambda x: 1 if x == 'Y' else 0)
data_encoded = pd.get_dummies(data.drop(columns=['_c39']))  # Drop non-informative column

# Separating the majority and minority classes
data_majority = data_encoded[data_encoded.fraud_reported == 0]
data_minority = data_encoded[data_encoded.fraud_reported == 1]

# Oversampling the minority class
data_minority_oversampled = resample(data_minority,
                                     replace=True,
                                     n_samples=len(data_majority),
                                     random_state=42)

# Combining the oversampled minority class with the majority class
data_oversampled = pd.concat([data_majority, data_minority_oversampled])

# Preparing the features and target variable for the balanced dataset
X_oversampled = data_oversampled.drop('fraud_reported', axis=1)
y_oversampled = data_oversampled['fraud_reported']

# Splitting the balanced dataset into training and test sets
X_train_oversampled, X_test_oversampled, y_train_oversampled, y_test_oversampled = train_test_split(X_oversampled, y_oversampled, test_size=0.2, random_state=42)

# Initialize the Logistic Regression model for the balanced dataset
log_reg_oversampled = LogisticRegression(max_iter=1000)

# Fit the model with the training data
log_reg_oversampled.fit(X_train_oversampled, y_train_oversampled)

# Predictions on the test set
y_pred_oversampled = log_reg_oversampled.predict(X_test_oversampled)

# Evaluation
classification_report_oversampled = classification_report(y_test_oversampled, y_pred_oversampled)
confusion_matrix_oversampled = confusion_matrix(y_test_oversampled, y_pred_oversampled)
print(classification_report_oversampled)
print("Confusion Matrix:\n",confusion_matrix_oversampled)
```

**Explanation:**

- **Recall:** The model is somewhat better at properly recognising fraudulent claims, as evidenced by the recall score of 0.62 for class 1 (fraudulent claims), which is marginally higher than the 0.52 score for class 0 (non-fraudulent claims). As a result of an imbalance in the model's capacity to detect both classes, the lower recall for non-fraudulent claims indicates that a sizable percentage of legitimate claims are missed.

- **F1-Score:** With a somewhat higher score for fraudulent claims, the F1-scores show a balanced performance between precision and recall, indicating that the model does marginally better at detecting fraud than non-fraudulent claims. Even while more optimisation could still enhance the model's overall classification performance, this suggests that it manages the trade-off between false positives and false negatives very well.

- **Precision:** Moderate accuracy in properly recognising positive predictions for both fraudulent and non-fraudulent claims is indicated by the model's precision score of 0.57 for both classes. This shows that there is potential for improvement in the model's ability to differentiate between the two classes, even though it may produce positive predictions. A sizable percentage of these forecasts may still be off.

- **Support:** shows how many cases there are in each class in the test set.


 **Confusion Matrix Interpretation:**

- **True Negatives:** The identification of 77 non-fraudulent claims was accurate.

- **True Positives**: The identification of 95 bogus claims was accurate.

- **False Positives**: 71 cases were misidentified as false claims.

- **False Negatives:** According to the model, 59 false claims were overlooked.

**Conclusion:**

- A modest degree of efficacy is indicated by an overall accuracy of 57%. The model does a better job of detecting fraudulent claims than non-fraudulent ones, but it still has to be improved in order to reliably distinguish between the two classes, as evidenced by the large percentage of false positives and false negatives.

# TEST 3: Undersampling the Majority Class:
